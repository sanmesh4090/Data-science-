# Array functions

```python
#Array functions
#ones
one=np.ones((3,4))
print(one)
```

```python
#zeros
zero=np.zeros((5,4))
print(zero)
```

```python
#full
full=np.full((4,5),10)
print(full)
```

```python
#random.rand
a=np.random.rand(4,4)
print(a)
```

```python
#random.randn
a=np.random.randn(3,4)
print(a)
```

```python
#random.randint
a=np.random.randint(1,10,(5,5))
print(a)
```

```python
#arrange
a=np.arange(1,11,2)
print(a)
```

```python
#eye
a=np.eye(5)
print(a)
```
