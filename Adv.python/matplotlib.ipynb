# linechart

```python
df["Month"] = pd.to_datetime(df["Date_Posted"]).dt.month_name()

print(df.groupby("Month")["Openings"].sum())
```

```python
monthly = df.groupby("Month")["Openings"].sum()

plt.figure(figsize=(10, 6))

monthly.plot(
    kind="line",
    marker="o",
    color="blue",
    linewidth=2,
    linestyle="--"
)

plt.xlabel("Months")
plt.ylabel("Openings")
plt.title("Monthly Openings")
plt.xticks(rotation=0)
plt.grid(True)

plt.tight_layout()
plt.savefig("Linechart.png")
plt.show()
```

```python
monthly = df.groupby('Month')['Openings'].sum()

plt.figure(figsize=(10, 6))

monthly.plot(
    kind="line",
    marker="o",
    color="blue",
    linewidth=2,
    linestyle="--"
)

plt.show()
```

# Bar chart

```python
job=df.groupby('Job_Title')['Openings'].sum().sort_values(ascending=False).head(10)
job.plot(kind='bar',color='purple',edgecolor='black')
plt.title('Job wise no of openings')
plt.xlabel('Job Title')
plt.ylabel('No of Openings')
plt.xticks(rotation=90)
plt.tight_layout()
plt.grid()
plt.savefig("BarChart.png")
plt.show()
```

# scatter plot

```python
plt.scatter(df['Salary_LPA'], df['Applicants'], color='red', edgecolor='black', marker='*',alpha=0.8)
plt.title('Scatterplot chart')
plt.xlabel('Applicants')
plt.ylabel('Salary_LPA')
plt.show()
```

# Histogram chart

```python
df.columns
```

```python
plt.hist(df['Company_Rating'], bins=11, edgecolor='black' ,color='purple')
plt.title('Histogram of Company Ratings')
plt.xlabel('Company Ratings')
plt.ylabel('Frequency')
plt.savefig("Histogram.png")
plt.show()
```

# Pie chart

```python
month=df.groupby('Month')['Openings'].sum()
month.plot(kind='pie',autopct='%1.2f%%',shadow="True")
plt.title('Monthly Openings')
plt.savefig("Piechart.png")
plt.show()
```
