```python
import pandas as pd
data = {
    "City": ["Pune", "Mumbai", "Nagpur", "Nashik", "Aurangabad", "Amravati"],
    "Orders": [120, 180, 90, 140, 110, 70],
    "Revenue": [4000, 3600, 1800, 2800, 2200, 1400]
}

df = pd.DataFrame(data)

df
```

```python
%pip install matplotlib
```

```python
#Line plot
import matplotlib.pyplot as plt
```

```python
plt.plot(df["City"],df["Revenue"],df["Orders"])
#Title
plt.title("Blinkit delivery app -- revenue by City")
#Labels
plt.xlabel("City")
plt.ylabel("Revenue")
#show the plot
plt.show()
```

```python
#Bar plot
plt.bar(df["City"],df["Orders"])
plt.title("Blinkit delivery app -- Orders by City")
plt.xlabel("City")
plt.ylabel("Orders")
plt.show()
```

```python
#Histogram
plt.hist(df["Revenue"])
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```
