```python
class car:
    def __init__(self,brand,model):
        self.__brand=brand
        self.model=model

    #getter method
    def get_brand(self):
        return f"{self.__brand}"

    def info(self):
        return f"{self.__brand} and the model is {self.model}"
```

```python
car1=car("Tata","Punch")
```

```python
car1.brand
```

```python
car1.get_brand()
```

```python
class Bank:
    def __init__(self,bank_name,balance):
        self.__bank_name=bank_name
        self.__balance=balance


    # getter method
    def get_bank_name(self):
        return self.__bank_name

    def get_balance(self):
        return self.__balance

    def info(self):
        return f"{self.__bank_name},balance:${self.__balance}"
```

```python
#object creation
bank1=Bank("SBI",10000)
```

```python
bank1.get_bank_name()
```

```python
bank1.get_balance()
```

```python
bank1.info()
```
