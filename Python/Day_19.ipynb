## Error and exceptions handling in python

```python
try:
    num=input()
    result=num/10
    print(result)

except Exception as e:
    print(e)
```

```python
try:
    num=input()
    div=num/10
    print(div)

except Exception as e:
    print("please enter the nuemeric input")
```

```python
try:
    num=input()
    div=num/10
    print(div)

except Exception as e:
    print("please enter the nuemeric input",e)
```

```python
try:
    age=int(input())
    if age>=18:
        print("you can apply for voting")

    else:
        print("you can wait for turn into 18")

except Exception as e:
    print("please enter nueneric only")
```

```python
try:
    my_list=[1,2,3]
    my_list[4]

except Exception as e:
    print("please enter the list index length")
```

```python
# 1.ZeroDivisionError

try:
    10/0

except ZeroDivisionError as z:
    print(z)
```

```python
# 2.ValueError

try:
    num=int("abc")

except ValueError as v:
    print(v)
```
