## Regex in python

```python
# use \d ->matches any digits(0-9)
import re
text="My phone number is 9988776655"

#search for digits
pattern=r"\d+"

result=re.findall(pattern,text)
print(result)




```

```python
# use \D -> matches any **non character**(anything that is not 0-9)
import re
text="My phone number is 9988776655"

pattern=r"\D+"

result=re.findall(pattern,text)
print(result)
```

```python
text="code123!"

pattern=r"\D+"
result=re.findall(pattern,text)
print(result)
```

```python
#w -> matches any word character (letters,digits,underscore)

text="python_3 is very powerfull tool!"

pattern=r"\w+"
result=re.findall(pattern,text)
print(result)
```

```python
#\W -> matches any non word characters (anything that is NOT a lettter,digit,underscore)

text="python_3 is great!"

pattern=r"\W+"
result=re.findall(pattern,text)
print(result)
```

```python
#\s -> matches any whitespace character(space,tab,newline)

text="python is fun to learn"

pattern=r"\s+"
result=re.findall(pattern,text)
print(result)

```

```python
#\S -> matches any non whitespace character (anything that is not a space,tab)


text="python is fun to learn"

pattern=r"\S+"
result=re.findall(pattern,text)
print(result)
```

```python
#. ->matches any single character(except newline '\n')

text="Data science is fun"

pattern=r"."

result=re.findall(pattern,text)
print(result)
```

```python
#^ -> matches the start of string (or start of a line in multiple mode)
text="Data science is fun"

pattern=r"^Data"

result=re.findall(pattern,text)
print(result)
```

```python
#$ -> matches the end of string (or end of line in multiple mode)
text="Regex is powerfull tool"

pattern=r"tool$"

result=re.findall(pattern,text)
print(result)
```

```python
# + -> matches one or more repitation 
text="rrr rocks really right!"

pattern=r"r+"

result=re.findall(pattern,text)
print(result)
```

```python
#Regex pattern for emails
text="Contact us at info@example.com or support@domain.org for help."
pattern=r"[A-Za-z0-0.%+-]+@[A-Za-z0-0.-]+\.[A-Za-z]{2,}"

result=re.findall(pattern,text)
print(result)
```
