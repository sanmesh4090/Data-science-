# .extend() - add the another list

```python
a=["milk","cookies","sugar"]
b=["butter","soap","salt"]

a.extend(b)
print(a)
```

```python
color1=["red","green","blue"]
color2=["yellow","orange","pink"]

color1.extend(color2)
print(color1)
```

# .Insert()-  add the element by the index position

```python
names=["sanmesh","sachin","sagar"]
names.insert(1,"Aniket")
print(names)
```

```python
names.insert(0,"Rohit")
print(names)
```

```python
numbers=[1,2,3,4,5]
numbers.insert(2,6)
print(numbers)
```

# .Pop()- used to pop the element

```python
cart=["shirt","pant","shoes","cap"]
print(cart)
cart.pop(2)
print(cart)
```

```python
names=["sanmesh","sachin","sagar"]
names.pop(1)
print(names)
```

```python
cart.pop(1)
cart.pop(0)
print(cart)
```

# .Clear() - used to empty list

```python
cart.clear()
print(cart)
```

```python
names.clear()
print(names)
```
