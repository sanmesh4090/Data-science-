```python
fruits = ("apple","banana","jackfruit")
```

```python
def show_fruits(fruits):
    for i in fruits:
        print(i)

show_fruits(fruits)
```

```python
def addition(a,b):
    return(a+b)
```

```python
addition(5,5)
```

```python
def addition(a,b,c):
    return(a+b+c)
```

```python
addition(5,6,5)
```

```python
def addition_5(a,b,c,d,e):
    return(a+b+c+d+e)

addition_5(5,5,6,8,5)
```
