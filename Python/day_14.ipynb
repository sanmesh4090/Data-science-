# Lambda Function

- can be write in single line
- when we want to quick code
- for the quick check
- it is a ananomyous 

```python
add = lambda a,b: a + b 
print(add(19,20))
```

```python
add = lambda a,b,c: a+b-c
print(add(5,6,5))
```

```python
greet = lambda: "hello world"
greet()
```

```python
def greet():
    return ("hello world")

greet()
```

```python
# write a fun to square 

def square(x):
    return(x*x)
```

```python
square(5)
```

```python
square = lambda x :x*x
```

```python
square(10)
```

```python
def num_check(num):
    if num % 2 == 0:
        print("even")
    else:
        print("odd")

num_check(1)
```

```python
num_check = lambda num : "even" if num % 2==0 else "odd"
```

```python
num_check(65)
```
