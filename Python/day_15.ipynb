```python
# crete an class 
class car:
    def __init__(self , brand , model , year): #attributes
        self.brand = brand #instance variables
        self.model = model #instance variables
        self.year = year #instance variables


# __init__ = constructor
# self = first parameter
```

```python
car1 = car("tata", "Punch" , 2025) #object
```

```python
car1.brand #calling attributes
```

```python
car1.model #calling attributes
```

```python
car1.year #calling attributes
```

```python
# class : Bank class

class Bank:
    def __init__(self, bank_name , branch):
        self.bank_name = bank_name
        self.branch = branch

    #method
    def info(self):
        return f"BANK IS : {self.bank_name}"
```

```python
# create an object

bank1 = Bank("state Bank Of India", "Main")
```

```python
bank1.bank_name
```

```python
bank1.branch
```

```python
#calling method
bank1.info()
```
