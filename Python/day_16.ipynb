# Inheritance 

```python
class vehicle: #parent class
    def __init__(self,brand):
        self.brand = brand

    #method
    def info(self):
        return f"Brand name is {self.brand}"

class car(vehicle): #child class
    def __init__(self,brand,model):
        # call parent constructor
        # super() - used to call a parent class constructor or methods 
        super().__init__(brand)
        self.model = model

    # method
    def info(self):
        return f"brand is {self.brand} and model is {self.model}" # polymorphism

    #method 
    # parent class method can call in child class but child class methods can't call in parent class
    def car_info(self):
        return f"brand is {self.model} and model is {self.model}"
```

```python
new_car = car('Tata','nexon')
new_car.info()
```

```python
new_car.car_info()
```

```python
old_car = vehicle("Tata")
```

```python
class vehicle: #parent class
    def __init__(self,brand):
        self.brand = brand

    #method
    def info(self):
        return f"Brand name is {self.brand}"

class car(vehicle): #child class
    def __init__(self,brand,model):
        # call parent constructor
        # super() - used to call a parent class constructor or methods 
        super().__init__(brand)
        self.model = model

    # method
    def info(self):
        return f"brand is {self.brand} and model is {self.model}" # polymorphism

```

```python
car_1 = vehicle("Tata") # parent class : vehicle
car_2 = car("tela", "model y") # child class :car
```

```python
car_1.info()
```

```python
car_2.info()
```

```python
#inheritance
#parent class
# info : return - bank name

class Bank:
    def __init__(self,bank_name):
        self.bank_name = bank_name

    #method
    def info(self):
        return f" bank name : {self.bank_name}"

# child class
class user(Bank):
    def __init__(self,bank_name,user_name):
        # call parent constructor
        super().__init__(bank_name)
        self.user_name = user_name

    def user_info(self):
        return f"user:{self.user_name} , Bank:{self.bank_name}"
```

```python
bank_1 = Bank("SBI") # creating object into parent class
```

```python
bank_1.info() #calling from parent class
```

```python
user_1 = user("SBI","rohit") #creating object into child class
```

```python
user_1.user_info() # calling from parent class and child class - it is called polymorphism
```

```python
# OOPs with  Apple device
# parent class - Hierarchical Inheritance

class Apple:
    def __init__(self, brand_name):
        self.brand_name = brand_name

    def info(self):
        return f"Brand: {self.brand_name}"

class iphone(Apple):
    def __init__(self,brand_name,model):
        super().__init__(brand_name)
        self.model = model

    def info(self):
        return f"Brand : {self.brand_name}, model: {self.model}"

# creating second child class 

class ipad(Apple):
    def __init__(self,brand_name,version):
        super().__init__(brand_name)
        self.version = version

    def info(self):
        return f"Brand : {self.brand_name}, version: {self.version}"
```

```python
apple_1 = Apple("iphone ") # creating a object in parent class
```

```python
apple_1.info() #calling method 
```

```python
iphone_1 = iphone("iphone","18") # creating a first object in child class
```

```python
iphone_1.info() # calling first child , it is called polymorphism
```

```python
ipad_1 = ipad("ipad_pro","5") # creating second object in parent class it is called Hierarchical Inheritance
```

```python
ipad_1.info() # calling second child class 
```
