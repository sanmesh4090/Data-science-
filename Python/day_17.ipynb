# abstract method

```python
from abc import ABC, abstractmethod
```

```python
# abstract parent class
class BankServer(ABC):
    def __init__(self,bank_name):
        self.bank_name = bank_name

    @abstractmethod
    def set_security(self):
        pass

```

```python
# child class Onlline banking

class OnlineBanking(BankServer):
    def __init__(self,bank_name,user_id):
        super().__init__(bank_name)
        self.user_id = user_id

    def set_security(self):
        return f"profile is secured with two factor Auth"
```

```python
class card(BankServer):
    def __init__(self, bank_name,card_num):
        super().__init__(bank_name)
        self.card_num = card_num

    def set_security(self):
        return f"{self.card_num} is secured with pin"
```

```python
online = OnlineBanking("SBI",9665596899) #object
```

```python
card_1 = card("SBI",9665596899) #object
```

```python
card_1.set_security() #calling method
```

```python
class MobileApp(BankServer):
    def __init__(self, bank_name, app_version):
        super().__init__(bank_name)
        self.app_version = app_version

    def set_security(self):
        return f"{ self.app_version} Is Secured With Fingerprint"
```

```python
app_1 = MobileApp("Bank of Python" , 3.11) #object
```
