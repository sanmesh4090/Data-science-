```python
def has_duplicates(data):
    if len(data) != len(set(data)):
        return "True"
    else:
        return "False"
    
print([1,2,2,3], "-", has_duplicates([1,2,2,3]))  
print([1,2,3,4], "-", has_duplicates([1,2,3,4]))  
```
