```python
#Abstraction
from abc import ABC,abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Dog barks"

```

```python
d=Dog()
```

```python
d.sound()
```

```python
#Encapsulation
class student:
    def __init__(self,name,marks):
        self.name=name
        self.__marks=marks

    def details(self):
        print("Name:",self.name)
        print("Marks:",self.__marks)

        
```

```python
s=student("Sanmesh",78)
```

```python
s.details()
```
